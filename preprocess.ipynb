```python
# read data
metro = pd.read_csv('annual_metropolitan_train_station_entries_fy_2023_2024.csv')
regional = pd.read_csv('annual_regional_train_station_entries_fy_2023_2024.csv')
```

```python
# keep relevant columns
metro = metro[['Stop_name', 'Stop_lat', 'Stop_long', 'Pax_annual']]
regional = regional[['Stop_name', 'Stop_lat', 'Stop_long', 'Pax_annual']]

# rename regional station names
regional['Stop_name'] = regional['Stop_name'].str.replace(r' Railway.*', '', regex=True).str.strip()

# combine dataframes
df = pd.concat([metro, regional], axis = 0)

# rename columns
df.columns = ['Name', 'Lat', 'Long', 'Patronage']

# combine southern cross data
sc = df[df['Name'] == 'Southern Cross']
new = pd.DataFrame([{'Name': 'Southern Cross', 'Lat': sc.iloc[0]['Lat'], 'Long': sc.iloc[0]['Long'], 'Patronage': sc['Patronage'].sum()}])
df = df[df['Name'] != 'Southern Cross']
df = pd.concat([df, new], axis = 0)

# correct station names
df.loc[df['Name'] == 'Mckinnon', 'Name'] = 'McKinnon'
df.loc[df['Name'] == 'Glenhuntly', 'Name'] = 'Glen Huntly'

# reset index and sort
df = df.sort_values(by = 'Name')
df = df.reset_index(drop = True)
```
