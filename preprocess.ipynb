```python
import pandas as pd
```

```python
# read data
metro = pd.read_csv('annual_metropolitan_train_station_entries_fy_2023_2024.csv')
regional = pd.read_csv('annual_regional_train_station_entries_fy_2023_2024.csv')
```

```python
# keep relevant columns
metro = metro[['Stop_name', 'Stop_lat', 'Stop_long', 'Pax_annual']]
regional = regional[['Stop_name', 'Stop_lat', 'Stop_long', 'Pax_annual']]

# rename regional station names
regional['Stop_name'] = regional['Stop_name'].str.replace(r' Railway.*', '', regex=True).str.strip()

# combine dataframes
df = pd.concat([metro, regional], axis = 0)

# rename columns
df.columns = ['Name', 'Lat', 'Long', 'Patronage']

# combine southern cross data
sc = df[df['Name'] == 'Southern Cross']
new = pd.DataFrame([{'Name': 'Southern Cross', 'Lat': sc.iloc[0]['Lat'], 'Long': sc.iloc[0]['Long'], 'Patronage': sc['Patronage'].sum()}])
df = df[df['Name'] != 'Southern Cross']
df = pd.concat([df, new], axis = 0)

# reset index and sort
df = df.sort_values(by = 'Name')
df = df.reset_index(drop = True)
```

```python
# manually add station type
interchange = ['Ballarat', 'Bendigo', 'Broadmeadows', 'Burnley', 'Camberwell', 'Caulfield', 'Clayton', 'Clifton Hill', 'Craigieburn', 'Dandenong', 'Flagstaff', 'Flinders Street', 'Footscray', 'Frankston', 'Laverton', 'Melbourne Central', 'Newport', 'North Melbourne', 'Pakenham', 'Parliament', 'Richmond', 'Ringwood', 'Seymour', 'South Yarra', 'Southern Cross', 'Sunbury', 'Sunshine', 'Watergardens']
staffed = ['Albion', 'Albury', 'Ararat', 'Armadale', 'Ascot Vale', 'Ashburton', 'Auburn', 'Bacchus Marsh', 'Bairnsdale', 'Balaclava', 'Ballan', 'Bayswater', 'Belgrave', 'Bell', 'Benalla', 'Bentleigh', 'Berwick', 'Blackburn', 'Boronia', 'Box Hill', 'Brighton Beach', 'Broadford', 'Camperdown', 'Carnegie', 'Caroline Springs', 'Carrum', 'Castlemaine', 'Cheltenham', 'Clarkefield', 'Cobblebank', 'Coburg', 'Colac', 'Cranbourne', 'Croydon', 'Darling', 'Deer Park', 'Drouin', 'East Malvern', 'Echuca', 'Elsternwick', 'Eltham', 'Epping', 'Essendon', 'Fairfield', 'Ferntree Gully', 'Geelong', 'Gisborne', 'Glen Huntly', 'Glen Waverley', 'Glenferrie', 'Glenroy', 'Gowrie', 'Greensborough', 'Hawkstowe', 'Heatherdale', 'Heidelberg', 'Holmesglen', 'Hoppers Crossing', 'Huntingdale', 'Hurstbridge', 'Ivanhoe', 'Jolimont', 'Kangaroo Flat', 'Kilmore East', 'Kyneton', 'Lara', 'Lilydale', 'Macleod', 'Malvern', 'Marshall', 'Maryborough', 'Melton', 'Mentone', 'Mernda', 'Middle Brighton', 'Middle Gorge', 'Mitcham', 'Moe', 'Moonee Ponds', 'Moorabbin', 'Mooroolbark', 'Mordialloc', 'Morwell', 'Mount Waverley', 'Murrumbeena', 'Narre Warren', 'Noble Park', 'North Brighton', 'North Geelong', 'Nunawading', 'Oakleigh', 'Ormond', 'Prahran', 'Preston', 'Rosanna', 'Reservoir', 'Riddells Creek', 'Ripponlea', 'Sale', 'Sandringham', 'Shepparton', 'South Geelong', 'South Morang', 'Springvale', 'St Albans', 'Swan Hill', 'Syndal', 'Tarneit', 'Thomastown', 'Tooronga', 'Traralgon', 'Union', 'Upfield', 'Upper Ferntree Gully', 'Wallan', 'Wangaratta', 'Warragul', 'Warrnambool', 'Watsonia', 'Waurn Ponds', 'Wendouree', 'Werribee', 'Westall', 'Williams Landing', 'Williamstown', 'Winchelsea', 'Windsor', 'Wodonga', 'Woodend', 'Wyndham Vale', 'Yarraville']

def stationtype(station):
    if station in interchange:
        return 'Interchange'
    elif station in staffed:
        return 'Staffed'
    else:
        return 'Unstaffed'
    
df['Type'] = df['Name'].apply(stationtype)
```
